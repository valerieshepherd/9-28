# Exercise 5 (10 points)

A farmer has 80 acres of land available for planting wheat and corn. Each acre of wheat yields a profit of $200 and requires 3 hours of labor. Each acre of corn yields a profit of $300 and requires 4 hours of labor. The farmer has at most 240 labor hours available during the planting season. Find the optimal number of acres to plant wheat and corn that maximizes the farmers profit.

```python
from scipy.optimize import linprog

c = [-200, -300]

A = [
    [1, 1],      # Land
    [3, 4]       # Labor
]

b = [80, 240]

result = linprog(c, A_ub=A, b_ub=b, bounds=[(0, None), (0, None)], method="highs")

print("Optimal acres of wheat:", result.x[0])
print("Optimal acres of corn:", result.x[1])
print("Maximum profit: $", -result.fun)
```

# Exercise 6 (10 points)

A person needs at least 500 calories and 30 grams of protein per meal.

- Food A: 250 calories, 10g protein, cost \$2
- Food B: 200 calories, 20g protein, cost \$3

Formulate and solve an LP to minimize the cost while meeting the nutritional requirements.

```python
from scipy.optimize import linprog

c = [2, 3]

A = [
    [-250, -200],  # Calories
    [-10, -20]     # Protein
]

b = [-500, -30]

result = linprog(
    c,
    A_ub=A,
    b_ub=b,
    bounds=[(0, None), (0, None)],
    method="highs"
)

print("Food A servings:", result.x[0])
print("Food B servings:", result.x[1])
print("Minimum cost: $", result.fun)
```

# Exercise 7 (5 points)

Nonlinear Programming (NLP) generalizes Linear Programming (LP) by allowing the objective function, some constraints, or both to be nonlinear functions of the decision variables. Explain the difference between a **convex NLP** and a **non-convex NLP**, and explain why this distinction matters when using a general-purpose NLP solver (e.g., `scipy.optimize.minimize`).

A convex nonlinear programming problem has a convex feasible region and a convex objective function when minimizing. For a convex problem, any local optimum is also a global optimum. This makes the problem easier for a general-purpose solver such as scipy.optimize.minimize because finding a local minimum can guarantee that the solution is globally optimal under the appropriate convexity conditions.

A non-convex NLP may have multiple local optima. A solver such as scipy.optimize.minimize can therefore find a local solution that is not necessarily the global optimum. The starting values and optimization method can affect the solution.

# Exercise 8 (10 points)

A company plans to allocate up to $80,000 (in thousands, 80) across two marketing channels: social media ($x_1$) and television ($x_2$). Historical data show that expected revenue from each channel exhibits diminishing returns and is well approximated by

$$\text{Revenue}(x_1, x_2) = 120\sqrt{x_1} + 90\sqrt{x_2} \quad \text{(in thousands of dollars)}.$$

Formulate and solve the resulting nonlinear program to determine the budget split that maximizes total revenue, subject to the budget constraint and non-negativity of $x_1$ and $x_2$. Use `scipy.optimize.minimize`.

```python
import numpy as np
from scipy.optimize import minimize

def revenue(x):
    x1, x2 = x
    return 120 * np.sqrt(x1) + 90 * np.sqrt(x2)

def objective(x):
    return -revenue(x)

constraints = [
    {
        'type': 'ineq',
        'fun': lambda x: 80 - x[0] - x[1]
    }
]

bounds = [
    (0, None),
    (0, None)
]

x0 = [40, 40]

result = minimize(
    objective,
    x0,
    method='SLSQP',
    bounds=bounds,
    constraints=constraints
)

print("Optimal social media budget (x1):", result.x[0])
print("Optimal television budget (x2):", result.x[1])
print("Maximum revenue ($ thousands):", -result.fun)
```

The optimal budget allocation is $44,000 to social media and $36,000 to television. This uses the entire $80,000 budget and produces a maximum expected revenue of approximately $1,335.96 thousand, or $1,335,960.
